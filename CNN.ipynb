# CNN (CONVOLUTIONAL NEURAL NETWORK)

## A Convolutional Neural Network (CNN) is a type of Deep Learning model that is specially designed to understand images.

## Instead of looking at the entire image at once, CNN looks at small parts of the image, learns simple features first, and gradually combines them to recognize complex objects.

### A CNN (convolutional neural network) is a neural network built specifically for grid-like data — mainly images. Instead of treating every pixel as an independent input like a plain ANN does, it uses small filters (kernels) that slide across the image and pick up local patterns — edges, textures, curves — then stack those patterns into more complex shapes as you go deeper.

## WHAT IS KERNELS ?

## A kernel (also called a filter) is a small matrix of numbers — usually 3×3 or 5×5 — that slides across the image and does one simple operation at each position: element-wise multiply, then sum.

### THIS SINGLE NUMBER BECOME'S THE OUPUT FOR THE FEATURE MAP

### ---KERNEL IS A 3 * 3 OR 5 * 5  MATRIX WHICH CONSITS OF SOME NUMBERS , IT STARTS FROM ONE PLACE OF THE IMAGE AND SLIDES(MOVE) ACROSS THE ALL POSITION OF THE IMAGE , AT EACH POSITION IT MULTIPLE'S WITH THE NUMBER PRESENT IN IMAGE AND SUMS IT UP AS **X** THIS SINGLE **X** IS THE OUTPUT OF SINGLE FRAME OF THE OUTPUT 


### --MORE WITH OPERATION DONE BY THE KERNEL ---> ELEMENT WISE MULTIPLICATION, THE FIRST ROW IS MULTIPLIED WITH FIRST ROW LIKE WISE IT GETS MULTIPLIED AND GETS SUMMED LATER (NOT MATRIX MULTIPLICATION)

### WHAT IS STRIDE ?

### ---NO OF PIXELS A FILTER/KERNEL AFTER EACH CONVOLUTION OPERATION

### ---STEP SIZE OF THE JUMP AFTER EACH OPERATION

### WHAT IS PADDING ?

### Padding is the process of adding extra pixels (usually zeros) around the border of an input image before applying the convolution operation.

### AS THE IMAGE SHRINKS AFTER THE CONVOLUTION OPERATION , WE MIGTH LOSS SOME IMPORTANT INFO AT THE OUTER PART OF IMAGE SO INORDER TO SORT IT OF , WE USE PADDING

## POOLING

### IT IS A DOWNSAMPLING TECHINUE WHICH IS APPLIED AFTER THE ACTIVATION FUNCTION IN A FEATURE MAP 

### --IT IS USED FOR DIMENSION REDUCTION(Meanwhile we can reduce the computational power) WHILE RETAINING THE MOST IMPORTANT INFORMATION,  CONTROLS OVERFITTING ALSO EXTRACT THE DOMINANT FEATURES

### Pooling is a downsampling operation applied to the feature map after activation, which reduces its spatial dimensions (height & width) while retaining the most important information, in order to reduce computation, control overfitting, and make the learned features more robust to small translations in the input.

## WHAT IS MAX POOLING ?

### IT IS A POOLING OPERTAION ,WHERE IN EACH WINDOW WE CHOOSE THE MAXIMUM VALUE AS THE OUTPUT 

### Max pooling is a operation where, for each window position on the feature map, you take the maximum value inside that window and discard the rest — keeping only the strongest activation from each region.

##  WHAT IS AVERAGE POOLING ?

### IN EVERY WINDOW WE CHOOSE THE OUTPUT AS THE **AVERAGE OF THE ELEMENTS** PRESENT INSIDE THE WINDOW

### AVERAGE POOLING is a pooling operation where, for each window position on the feature map, you take the average (mean) of all values inside that window instead of the max.
