# RNN --( RECURRENT NUERAL NETWORK)

## Sequential data = data arranged in a meaningful order, where previous observations can provide information about the next observation.


## Recurrent Neural Networks (RNNs) are a class of neural networks designed to process sequential data by retaining information from previous steps. They are especially effective for tasks where context and order matter.

### Imagine reading a sentence and you try to predict the next word, you don’t rely only on the current word but also remember the words that came before. RNNs work similarly by “remembering” past information i.e it considers all the earlier words to choose the most likely next word

### Its application are text summarization ,translation ,language modelling , sentimental analysis 

### Normal Neural Network doenst works with sequential data, While RNN works on it 

### RNN passes the information from the previous step to the next step

# **BPTT — Backpropagation Through Time**

## BPTT is normal backpropagation applied to an RNN after we unroll the RNN across time steps.

## then what does the thorugh time means ?


 ### --because rnn hidden state carries inofrmation from prev step to current step

#### --AS SAME WIEGTHS ARE REUSED AND HIDDEN STATE IS DEPENDENT , WE CANT DO NORMAL BACK PROP SOO WE USE HERE BPTT

## GENARAL DEFNTION OF BPTT

### In an RNN, the same weights are used repeatedly across time, so BPTT calculates how those shared weights contributed to the error at every time step.

### SOO INTIALLY WE WILL DO A FORWARD PROP , FROM IT WE WILLL CALCULATE THE LOSS FUNCTION

### SOO AGAIN WE GO BACK CALCULATE GRADIENTS FOR THE SHARED WEIGTHS , AND UPDATE THE WEIGTHS 

### AND REDUCE THE LOSS FUNCTION

### HERE WHILE WE CALCUATE THE GRADIENT'S WE ALSO CONDIER THE TEMPRORAL DEPEDENCIES(CURRENT STATE IS DEPENTED ON PREV STATE)

## ALGO OF BPTT

### 1. Forward propagate through all time steps
   #### ---x1 → h1 → y1
   #### ---x2 → h2 → y2
   #### ---x3 → h3 → y3

### 2. Calculate the total loss
#### ---L = L1 + L2 + L3

### 3. Backpropagate from the last time step to the first
   #### ---L3 → h3 → h2 → h1

   #### ---This is important because:
   #### ---h3 depends on h2
   #### ---h2 depends on h1

### 4. Calculate gradients while following these temporal dependencies. ,shared weigths  

### 5. Since the same weights are reused at every time step,
   #### ---accumulate the gradients contributed by each time step.

### 6. Update the shared weights using the accumulated gradient.

### 7. This reduces the loss in the next iteration.

# **Vanishing Gradient**

## Vanishing gradient is a problem where the gradients become extremely small as they are propagated backward through many time steps, causing the RNN to learn very slowly or fail to learn long-term dependencies.

### The gradients become extremely small (approaching zero), so the corresponding weights receive almost no update and learning of long-term dependencies becomes very difficult.

### as we do have many time step , while we go back during back prop , the gradients become smaller , which stops the learning 

### BPTT can lead to the vanishing-gradient problem because, when gradients are propagated backward through many RNN time steps, they are repeatedly multiplied by derivatives. If those values are small, the gradient becomes extremely small.
