# RNN --( RECURRENT NUERAL NETWORK)

## Sequential data = data arranged in a meaningful order, where previous observations can provide information about the next observation.


## Recurrent Neural Networks (RNNs) are a class of neural networks designed to process sequential data by retaining information from previous steps. They are especially effective for tasks where context and order matter.

### Imagine reading a sentence and you try to predict the next word, you don’t rely only on the current word but also remember the words that came before. RNNs work similarly by “remembering” past information i.e it considers all the earlier words to choose the most likely next word

### Its application are text summarization ,translation ,language modelling , sentimental analysis 

### Normal Neural Network doenst works with sequential data, While RNN works on it 

### RNN passes the information from the previous step to the next step

## 2) RNN UNFOLDING 

## RNN unfolding is the process of expanding an RNN into multiple time steps to show how it processes sequential data and passes information from one time step to the next.

### THE SAME RECURRENT NEURON IS APLLIED TO MULTIPLE TIME STEPS , TO SHOW HOW IT PROCESS THE SEQUENTIAL DATA AND PASSES INFORMATION FROM ONE TIME STEP TO ANOTHER TIME STEP 

### THIS UNROLLING ENABLES BACKPROPOGATION THROUGH TIME (BPTT) WHICH WILL BE LEARNED LATER 

## **RNN ARCHITECTURE**

### WE ARE SEEING WHAT'S HAPPENING INSIDE OR ON A SINGLE NEURON 

### FIRST BEFORE GOING INTO RNN 

### --THE NORMAL NEURAL NETWORK , TREATS EVERY INPUT AS INDEPENDENT AND HAS NO MEMORY ON PREVIOUS STUFF 

### RNN AT EACH STEP , TAKE'S THE INPUT + THE INFORMATION FROM THE PREVIOUS STEP'S AND PRODUCES THE NEW HIDDEN STATE 

### CURRENT INPUT + PREV MEMORY ---> NEW MEMORY 

## Ht=sigma(Ux+Wh(t-1)+B)

 ### WHERE **H** IS NEW MEMORY , **X** IS THE INPUT , **H(t-1)** prev memory ,**B** IS BIAS AND **U** IS THE WEIGTHS FOR INPUT WHILE THE **W** IS THE WEIGTHS FOR PREV INPUT 

### THE WEIGTHS WILL BE SAME FOR EACH TIME STEP'S

## RNN USES THE SHARED WIEGTHS AT EACH TIME STEP'S

### THE OUTPUT OF THE RECURRENT NODE WILL CALCULATED BY APLLYING AN ACTIVATION FUNCTION TO THE WEIGTHED STATE 

### OUTPUT ---> ACTIVATION FUNCTION + HIDDEN STATE 

### Y=O(V⋅h+C) 

### -- WHERE **O** IS AN ACTIVATION FUCNTION , **H** IS THE HIDDEN STATE , **V** IS WEIGTHS AND **C** BIAS 

## NOTE 

### ---THE UNFOLDING ARE DONE IN A SINGLE NEURON , LIKE WISE IN A LAYER OF 5 NERUONS WE WILL DO 5 DIFFRENT UNFOLDING FOR THE 5 NEURONS ON EACH TIME STEP 

## ALGO OF RNN IN FORWARD PROP

### SOME IMPORTANT  POINTS BEFORE MOVING IN 

### ---CURRENT INPUT + PREV MEMORY --> NEW MEMORY 

### ---THE INPUT WILL BE BASED ON THE TIME STEP , I'E WE GET N INPUTS IN N TIME STEP -- SO NO CONNECTION BETWEEN THE OUTPUT OF THE PREV TIME TO INPUT OF THE NEW TIME STEP , INPUT WILL BE GIVEN IN EACH TIME STEP 

### --OUTPUT OF THE CURRENT NODE IS **NOT** THE EITHER MEMORY OR NEW MEMORY OF THE NEXT NODE 

## DEFINITION  OF SOME IMPORTANT TERMS

### NEW MEMORY ----> CURRENT INPUT + PREV MEMORY


### OUTPUT AT A TIME STEP -----> LINEAR TRANSFORMATION OF NEW MEMORY + ACTIVATION FUNCTION OF IT  

## NOW GOING INTO ALGO

### LETS TAKE A RECURRENT NODE 
### FIRST AT A TIME STEP WE GET A INPUT **X** AND WE DO WE A PREV HIDDEN STATE **H(t-1)** 

### WITH THIS **X** AND PREV MEMORY **H(t-1)** WE CAN GET THE NEW MEMORY **H(T)** (current input + prev memory)

### **OUTPUT** --> NOW WITH THIS NEW MEMORY , IF WE DO LINEAR TRANFORMATION (MULTIPLE WEIGTHS AND ADD BIAS ) AND USE AN ACTIVATION FUNCTION WE GET OUTPUT OF THE SINGLE TIME STEP 

### NOTE : THIS **OUTPUT** IS **NOT** THE **HIDDEN STATE** FOR NEXT , IT IS JUST AS OUTPUT , NO CONNETION OF IT WITH THE NEXT NODE 

### THE **NEW MEMORY** WILL BE THE HIDDEN STATE FOR THE NEXT TIME STEP , IT IS ONLY CONNECTED WITH NEXT TIME STEP WHILE THE OUTPUTS ARE INDEPENDENT 

### THE ACTIVATION FUNCTION USED HERE IS **tanH** , THIS CONVERTS THE OUTPUT INTO THE RANGE (-1 TO 1)

### WHY TANH ACTIVATION FUNCTION --->Zero-centered, allows memory to increase or decrease, strong gradients near 0

### NOW HOW DOES WE GET THE OUTPUT FROM THE UNFOLDING OF THE NODE 

### AS WE HAVE N TIMES STEP'S WE DO HAVE N OUTPUTS 

### SOO THE OUTPUT IS BASED ON THE RNN TYPE 

## IF **MANY TO MANY** (Aligned)

### Every input produces its own corresponding output. Input length = output length, and they're aligned position-by-position.

### SOO ALL THE OUTPUT'S AT THE TIME STEPS ARE TAKEN

## IF **MANY TO MANY** (Seq2Seq)

### The output sequence can be a completely different length than the input, and generation only starts after the encoder has read the entire input.

## IF **MANY TO ONE**

### Many inputs go in, only one output comes out — at the very end.

### WE TAKE THE ONLY THE OUTPUT AT THE END
