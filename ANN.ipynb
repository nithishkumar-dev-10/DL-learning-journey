## DROP OUT 

### --WHILE TRAING A NEURAL NETWROK ON SAME DATA FOR SEVERAL TIME'S SOME NEURONS BECOMES OVERLY DEPENDENT ON EACH OTHER , WHILE SOME SET OF NEARUONS TEAM UP'S AND MEMORIZSE THE DATA INSTEAD OF UNDERSTANDING THE DATA , WHICH LEADS TO OVERFITTING

## --Dropout is a regularization technique where, at every training step, each neuron in a layer is randomly and temporarily deactivated with probability p — meaning its output is set to zero and it contributes nothing to that step's forward and backward pass. The neuron's weights are untouched and it becomes active again in later steps, since the "drop" decision is re-randomized fresh every single batch. At test time, dropout is turned off completely and all neurons are used together.

### ---WE TEMPROIALY TURN OFF SOME NEURONS RADOMILY AT EACH STEP AND IN THE LATER STEPS WE ACTIVATE IT BACK AND DEACTIVE SOME OTHER SET OF NEURONS , INORDER TO REDUCE OVERFITTING (MEMORIZE THE DATA) 

### --AT THE DEACTIVATED STAGE THE NEURONS WEIGTHS ARE UNTOUCHED AND CONTRIBUTES NOTHING TO THE OUPUT (WE JUST LITREALY REMOVE THE NEURON AT A PARTICULAR STEP)

### --THE DROP OUT NEURONS ARE CHOOSEN RANDOMLY 

### WHAT DOES A PROBALITLY MEAN HERE ? 

### HERE , IT DENOTES HOW MUCH PROBABLITY EACH NEURON HAVE TO GET DROPPED OUT 

### --- NOTE : PROBALITY  DOESNT MEAN P(X) AMOUNT OF NEURON GET'S ELIMINATED , IT MEANS HOW MUCH EACH NURONS HAVE A CHANCE TO GET ELMINATE

## EARLY STOPPING

### --WE RUN THE NEAURAL NETWORK FOR X (EPOCH)

### --WE CALCULATE THE LOSS AFTER EACH EPOCH , SO THIS METHOD STOP THE NEURAL NETWORK ONCE THE LOSS FUNCION STOPS IMPROVING

#### --RETRAININIG THE NETWEOK WHEN THE LOSS NEVER'S IMPORVES MAKE THE MODEL OVERFITTING BY MEMORISING THE DATA , SO WE STOP THE EPOCH ONCE WHEN THE LOSS STOP'S TO IMPROVE 

### --Early stopping is a regularization technique where, during iterative training, we monitor the model's loss on a separate validation set after every epoch. Training is halted once the validation loss stops decreasing (and starts increasing) even if the training loss continues to fall — because that divergence indicates the model has begun memorizing the training data (overfitting) instead of learning generalizable patterns. This prevents overfitting without modifying the loss function itself, unlike L1/L2 regularization.
