## ARTIFICIAL NEURAL NETWORK (ANN)

### --An Artificial Neural Network (ANN) is a computational model inspired by the human brain.

### --IT CONSIST OF MANY INTERCONNECTED NEURONS WHICH IS CONNECTED EACH OTHER AND COMMUNICATES 

### --INSTEAD OF FOLLOWING PRE-DEFINED RULES ,IT LEARNS THE PATTERN DIRECTLY FROM THE DATA

### WHATS IS NEURONS 

### --- IT IS A SMALLEST COMPUTATIONAL  UNIT OF ANN 

### -- IT RECIVE'S THE INPUT DO SOME MATHEMATICAL STUFFS AND PRODUCE'S AND OUTPUT 

### CONNECTION'S

### --THE LINK BETWEEN TWO NODES THAT CARRIES INFORMATION

## WEIGTH'S

### --THESE ARE NUMERICAL VALUES WHICH DETERMINE HOW IMPORTANT EACH FEATURE IS ?


### WHY IT MATTERS ?

### --- NOT EVERY FEATURE HAVE EQUAL IMPORTANCE , WEIGTH TELLS THE NEURON WHICH FEATURE SHOULD INFLUENCE MORE 

## BIAS 

### --IT IS AN ADDITIONAL VALUES WHICH IS ADDED TO THE NEURON WHICH HELPS THE OUTPUT OF THE MODEL TO SHIFT 

### -- THIS MAKES THE MODEL MORE FLEXIBLE AND MAKES THE MODEL CAPABLE TO LEARN COMPLEX PATTERNS

## --Bias is a trainable parameter in an artificial neuron that is added to the weighted sum of the inputs before applying the activation function. It shifts the activation and enables the neuron to fit the data more accurately.

### BIAS IS APPLIED BEFORE THE FORWARD PROPOGATION

## ACTIVATION FUNCTION 

### --IT IS AN MATHEMATICAL FUNCTION WHICH IS APPLIED TO THE WEIGTHED SUM OF NEURON INPUT TO DETERMINE ITS OUTPUT 

### ---TAKES THE INPUT OF THE WIEGTHED SUMS , DO SOME MATHEMATICAL STUFF AND PRODUCE AND OUTPUT 

### WORKFLOW OF A SINGLE NODE

### INPUT --> MULTIPLY RANDOM WEIGTHS  ----> WEIGTHED SUMS --->ADD BIAS---> ACTIVATION FUNCTION ---> NEURON OUTPUT

## LAYERS IN NEURAL NETWORK

## INPUT LAYER 

### --THIS IS THE FIRST LAYER OF THE NETWORK , IT RECIEVES THE INPUT FROM THE DATA AND PASSES IT WITHOUT PERFORMING ANY MATHEMATICAL CALCULATION

### --IT IS THE ENTRY POINT, IT JUST RECIEVES THE INPUT FROM THE DATA AND PASSES IT INTO THE FIRST HIDDEN LAYER 

## HIDDEN LAYER 

### --IT IS THE LAYER BETWEEN THE INPUT AND THE OUTPUT LAYER , IT PERFORMS THE MATHEMATICAL FUNCTIONS AND LEARNS THE COMPLEX PATTERNS FROM THE INPUT LAYER 

### --THERE IS NO FIXED NUMBER FOR HIDDEN LAYER IT DEPENDS UPON THE COMPLEXITY OF THE PROBLEM 

### ---IT IS THE GROUP OF NEURONS 

### --THEN THE WORKFLOW IS ALSO SAME 

### --EACH NEURON GETS ALL INPUT FEATURES , MULTIPLYS ITS OWN WEIGTHS ,PRODUCE THE WEIGTHED SUM AND ADDS THE BIAS , AND APPLY THE ACTIVATION FUNCTION

## OUTPUT LAYER 

### --IT IS THE FINAL LAYER OF THE ANN

### -- IT RECIVES THE OUTPUT OF THE LAST HIDDEN LAYER AND PERFORMS THE SAME NEURON COMPUTATION AND GIVES THE FINAL OUTPUT 

### --THE WOKFLOW IS SAME AS THE HIDDEN LAYER BUT IT THE LAST LAYER IN ANN
